```python
# utils/embeddings.py
import ollama

def embed_text(text, model="nomic-embed-text"):
    """
    Generate embeddings for a given text using Ollama.
    Returns a list of floats (embedding vector).
    """
    try:
        response = ollama.embeddings(model=model, prompt=text)
        print(f'_________________________ embeding : {response['embedding']}')
        return response['embedding']
    except Exception as e:
        print(f"Embedding error: {e}")
        return None
```

```python
res = embed_text('my name is rushal')
```
